```python
import pandas as pd
import matplotlib.pyplot as plt

plt.Figure()
data= pd.read_csv("../files/input/news.csv", sep=",", index_col=0)
for col in data.columns:
    plt.plot(data[col], label=col)
plt.title("How people get their news", fontsize=16)
plt.legend(loc="upper right")
plt.show()
```

```python
plt.Figure()
    
colors={
        'Television':'dimgray',
        'Newspaper':'grey',
        'Internet':'tab:blue',
        'Radio':'lightgrey'
    }
    
data= pd.read_csv("../files/input/news.csv", sep=",", index_col=0)
for col in data.columns:
        plt.plot(data[col],
                 color=colors[col] ,
                 label=col)
plt.title("How people get their news", fontsize=16)
    
plt.show()
```

```python
plt.Figure()
    
colors={
    'Television':'dimgray',
    'Newspaper':'grey',
    'Internet':'tab:blue',
    'Radio':'lightgrey'
}

zorder={
    'Television':1,
    'Newspaper':1,
    'Internet':2,
    'Radio':1,
}

linewidths={
    'Television':2,
    'Newspaper':2,
    'Internet':4,
    'Radio':2,
}

data= pd.read_csv("../files/input/news.csv", sep=",", index_col=0)
for col in data.columns:
    plt.plot(data[col],
                color=colors[col] ,
                label=col,
                zorder=zorder[col],
                linewidth=linewidths[col],
                )
plt.title("How people get their news", fontsize=16)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['left'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().axes.get_yaxis().set_visible(False)

for col in data.columns:
    first_year=data.index[0]
    plt.scatter(
        x=first_year,
        y=data[col][first_year],
        color=colors[col],
        zorder=zorder[col],
    )
    last_year=data.index[-1]
    plt.scatter(
        x=last_year,
        y=data[col][last_year],
        color=colors[col],
        zorder=zorder[col],
    )


plt.show()
```
